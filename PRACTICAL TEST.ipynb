```python
from datetime import datetime, timedelta

# 1. Leap year
def is_leap_year(year):
    if year % 400 == 0:
        return True
    elif year % 100 == 0:
        return False
    elif year % 4 == 0:
        return True
    else:
        return False

for y in [1700, 2016, 2026, 2200, 2032]:
    print(y, is_leap_year(y))

# 2. Day of week
def day_of_week(text):
    d = datetime.strptime(text, "%Y-%m-%d")
    return d.strftime("%A")

for t in ['2026-05-01', '2026-10-02', '2027-03-14']:
    print(t, day_of_week(t))

# 3. Days between two dates
d1 = datetime(2026, 3, 10)
d2 = datetime(2026, 9, 20)
print((d2 - d1).days)

# 4. Formatting
now = datetime(2026, 11, 2, 21, 45, 10)
print(now.strftime("%d-%m-%Y"), now.strftime("%m/%d/%Y"))
print(now.strftime("%H:%M:%S"), now.strftime("%I:%M:%S %p"))

# 5. timedelta
later = now + timedelta(weeks=1, days=6, hours=3)
print(later.strftime("%Y-%m-%d %H:%M:%S"))
```

```python
# ---------- Q2: math and random ----------

# 1. Buses needed
print(math.ceil(129 / 25))

# 2. Seconds to h, m, s
total = 7384
h = total // 3600
m = (total % 3600) // 60
s = total % 60
print(h, m, s)

# 3. simplify using gcd
def simplify(n, d):
    g = math.gcd(n, d)
    return (n // g, d // g)

print(simplify(36, 84))

# 4. Tower height
dist = 120
angle = math.radians(60)
print(round(dist * math.tan(angle), 2))

# 5. Dice rolls with seed
random.seed(31)
rolls = [random.randint(1, 6) for _ in range(5)]
print(rolls, sum(rolls))

# 6. Shuffle and choice (no re-seed)
deck = list(range(1, 9))
random.shuffle(deck)
print(deck)
print(random.choice(["rock", "paper", "scissors"]))
```

```python
import os
import shutil

base = "store_data"
logs = os.path.join(base, "logs")
archive = os.path.join(base, "archive")

# 1. Create folders (second call must not raise error)
os.makedirs(logs, exist_ok=True)
os.makedirs(logs, exist_ok=True)

# 2. Write three files
with open(os.path.join(logs, "p.txt"), "w") as f:
    f.write("red\ngreen\n")

with open(os.path.join(logs, "q.txt"), "w") as f:
    f.write("blue\n")

with open(os.path.join(logs, "r.csv"), "w") as f:
    f.write("x,y\n1,2\n")

# 3. Sorted list of .txt files in logs
txt_files = sorted(name for name in os.listdir(logs) if name.endswith(".txt"))
print(txt_files)

# 4. Read p.txt: number of lines and first line
with open(os.path.join(logs, "p.txt")) as f:
    lines = f.read().splitlines()
print(len(lines), lines[0])

# 5. Create archive, move p.txt into it
os.makedirs(archive)
shutil.move(os.path.join(logs, "p.txt"), archive)
print(sorted(os.listdir(logs)), os.listdir(archive))

# 6. Rename q.txt to old_q.txt
os.rename(os.path.join(logs, "q.txt"), os.path.join(logs, "old_q.txt"))
print(os.path.exists(os.path.join(logs, "q.txt")),
      os.path.isfile(os.path.join(logs, "old_q.txt")))

# 7. Delete everything
shutil.rmtree(base)
print(os.path.exists(base))
```

```python
# ---------- Q8: masks, np.where, views ----------

# 1. Boolean masks on ages
ages = np.array([21, 34, 17, 29, 45, 16, 38])
print(ages[ages >= 18])
print(ages[(ages > 22) & (ages < 40)])
print(ages[~(ages >= 18)])

# 2. np.where for Pass/Fail
scores = np.array([82, 64, 77, 91, 58, 70])
print(np.where(scores >= 65, "Pass", "Fail"))

# 3. Replace masked values
temps = np.array([24, 38, 33, 41, 29, 36])
mask = temps > 35
temps[mask] = 36
print(temps, int(mask.sum()))

# 4. Block of rows 1-3, columns 2-4
mat = np.arange(6, 31).reshape(5, 5)
print(mat[1:4, 2:5])

# 5. View vs copy
# mat[0] is a view (shares memory with mat), so changing it changes mat;
# .copy() makes independent data, so changing the copy leaves mat unchanged.
view = mat[0]
view[0] = 99
print(mat[0, 0])

copy = mat[1].copy()
copy[0] = -1
print(mat[1, 0])
```

```python
# ---------- Q9: array maths and statistics ----------

# 1. Element-wise ops and matrix product
A = np.arange(6, 15).reshape(3, 3)
B = np.arange(11, 20).reshape(3, 3)
print(A * B)
print(np.round(A / B, 2))
print(A @ B)

# 2. z-score
arr = np.array([2, 4, 6, 8, 10], dtype=float)
z = (arr - arr.mean()) / arr.std()
print(np.round(z, 2), round(float(z.mean()), 6))

# 3. Euclidean distance
p = np.array([2, 3, 4])
q = np.array([5, 7, 16])
print(float(np.sqrt(np.sum((p - q) ** 2))))

# 4. Statistics
v = np.array([12, 5, 21, 8, 17, 3])
print(np.argmax(v), np.argmin(v), v.max(), v.min())
print(float(np.mean(v)), float(np.median(v)),
      round(float(np.var(v)), 2), round(float(np.std(v)), 2))

# 5. Sorting along axes
m = np.array([[9, 3, 1], [6, 8, 2], [5, 7, 4]])
print(np.sort(m, axis=0))
print(np.sort(m, axis=1))

# 6. Positive, negative, zero counts
pn = np.array([0, -5, 7, 0, -2, -9, 6, 1, 0])
print(int((pn > 0).sum()), int((pn < 0).sum()), int((pn == 0).sum()))
```

```python
# ---------- Q10: reshape, broadcasting, splitting ----------

# 1. reshape and flatten
vals = np.arange(1, 16)
print(vals.reshape(5, -1).shape)
print(vals.reshape(3, 5).flatten().tolist() == vals.tolist())

# 2. broadcasting
sales = np.array([[40, 15], [25, 35], [18, 22]])
bonus = np.array([3, 6, 2])[:, np.newaxis]
print(sales + bonus)

rates = np.array([0.15, 0.08])
print(np.round(sales * rates, 2))

# 3. split vs array_split
arr = np.arange(1, 13 + 1)
try:
    np.split(arr, 3)
except ValueError:
    print("ValueError")
print([part.tolist() for part in np.array_split(arr, 3)])
```

```python
import pandas as pd

# 1. Series
s = pd.Series(["Agent_1", "Agent_2", "Agent_3"], index=["a", "b", "c"])
print(s["b"], s.iloc[2], len(s))

# 2. DataFrame from list of dictionaries
items = pd.DataFrame([
    {"item": "pen", "qty": 10},
    {"item": "book", "qty": 4},
    {"item": "bag", "qty": 2},
])
print(items)

# 3. Read agents.csv
df = pd.read_csv("agents.csv")
print(df.head(3))
print(df.shape, list(df.columns))

# 4. Duplicates and missing values
print(df.duplicated().sum())
print(df.isna().sum())
```

```python
# 1. Remove duplicate rows
df = df.drop_duplicates()

# 2. Incentive: text -> numbers, missing -> 0, then int
df["Incentive"] = pd.to_numeric(df["Incentive"], errors="coerce")
df["Incentive"] = df["Incentive"].fillna(0).astype(int)

# 3. Pay: fill missing with rounded mean, then int
fill = int(round(df["Pay"].mean()))
df["Pay"] = df["Pay"].fillna(fill).astype(int)

# 4. StartDate -> datetime, plus Year, Weekday, Years
df["StartDate"] = pd.to_datetime(df["StartDate"])
df["Year"] = df["StartDate"].dt.year
df["Weekday"] = df["StartDate"].dt.day_name()
df["Years"] = (pd.Timestamp("2026-01-01") - df["StartDate"]).dt.days // 365

# 5. Output
print(fill, df.shape, df["Pay"].dtype, df["Incentive"].dtype)
print(df[["AgentID", "Year", "Weekday", "Years"]].head(4))
```

```python
# 1. Total column
df["Total"] = df["Pay"] + df["Incentive"]

# 2. South zone and Total > 31000
print(df[(df["Zone"] == "South") & (df["Total"] > 31000)][["AgentID", "Name", "Total"]])

# 3. Types and label/position access
print(type(df["Name"]).__name__, type(df[["Name"]]).__name__)
print(df.loc[0, "Name"], df.iloc[1, 1])

# 4. Band with apply, then ordered Categorical
def band(n):
    if n >= 36000:
        return "High"
    elif n >= 30000:
        return "Mid"
    else:
        return "Low"

df["Band"] = df["Total"].apply(band)
df["Band"] = pd.Categorical(df["Band"], categories=["Low", "Mid", "High"], ordered=True)

# 5. Sort and count
top = df.sort_values(["Band", "AgentID"], ascending=[False, True])
print(top[["AgentID", "Band"]].head(3))
print(df["Band"].value_counts().sort_index())
```

```python
# 1. Group by Zone: mean, max, count of Pay
g = df.groupby("Zone")["Pay"].agg(["mean", "max", "count"])
g["mean"] = g["mean"].round(1)
print(g.sort_values("mean", ascending=False))

# 2. Percentage of group total
df["PctOfGroup"] = (df["Pay"] / df.groupby("Zone")["Pay"].transform("sum") * 100).round(1)
print(df[["AgentID", "Zone", "PctOfGroup"]].head(4))

# 3. Pivot table
pt = pd.pivot_table(df, index="Zone", columns="City", values="Pay",
                    aggfunc="sum", fill_value=0)
print(pt)
```

```python
# 1. Left merge with supervisors
sup = pd.read_csv("supervisors.csv")
left = df.merge(sup, on="Zone", how="left")
print(left["Supervisor"].isna().sum(), left.shape)

# 2. Inner and outer merge
inner = df.merge(sup, on="Zone", how="inner")
outer = df.merge(sup, on="Zone", how="outer")
print(len(inner), len(outer))

# 3. Index, loc, reset_index, rename
emp = df.set_index("AgentID")
print(emp.loc[409, "Name"])
emp = emp.reset_index().rename(columns={"Pay": "Monthly Pay"})
print(list(emp.columns[:6]))

# 4. JSON save and read back
df[["AgentID", "Name"]].to_json("names.json", orient="records")
back = pd.read_json("names.json")
print(back.shape)
```
